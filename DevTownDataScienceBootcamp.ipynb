```python
from sklearn.model_selection import train_test_split # to train and test model
from sklearn.linear_model import LogisticRegression # classification technique
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
# computing null values fpr embark and deck parameter
df["embarked"].fillna("S", inplace=True)
df["deck"].fillna("Unknown", inplace=True)
```

```python
# remove uncessary cols that are not required for our model
df.drop(['class', 'who', 'adult_male', 'deck', 'embark_town', 'alive'], axis=1, inplace=True)
```

```python
# converting categorical data to numerical data
df["sex"] = df["sex"].map({'male':0, 'female':1})
df["embarked"] = df["embarked"].map({'S':0, 'C':1, 'Q':2})
df["alone"] = df["alone"].map({False:0, True:1})
df.fillna(0, inplace=True) # if any null values are left
```

Getting started with our Machine learning Model i.e., Supervised Learning using Logistic Regression

```python
# defining inputs and outputs
x = df.drop("survived", axis=1) # input
# from the data frame, remove survived col
# removed 'survived' from input cols instead of defining a whole list of cols
# now except for survive feature everything becomes input
# axis=1 means row wise and axis=0 means col wise
y = df["survived"] # output
```

```python
# train and test model
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
# test_size=0.2: take 20% of data as test data
# random_state=42: take random data for test and train rather than consective data
```

```python
# apply classification technique
model = LogisticRegression(max_iter=1000)
model.fit(x_train, y_train) # make sure that there are no null values
```

```python
# Predict the output
y_pred = model.predict(x_test)
```

```python
# Evalution: Check Accuracy and Confusion Matrix
print("Accuracy: ", accuracy_score(y_test, y_pred)) # accuracy can differ since we are using random_state

# we can use standard scalar ftn to standardize all the values from 0-1 to improve accuracy
```

```python
cm = confusion_matrix(y_test, y_pred) # create confusion matrix using predicted and actual value
sns.heatmap(cm, annot=True, fmt="d")
# annot=True means show values inside blocks 
# fmd="d" means show integer values
plt.title("Confusion Matrix")
plt.show()

# What does this confusion matrix show?
# 0: Not survived, 1: Survived
# 0,0: there were 0 people who didn't survive and predicted to be not survived (True Negative)
# 1,1: there were 31 people who survived and were predicted to be survived (True Positive)
# 0,1: 7 people died but they were predicted as not dead (False Positive)
# 1,0: 3 people survived but they were predicted died (False Negative)
```
